```python

from pyspark.sql.functions import col, when, round as sround, months_between

silver = spark.table("silver_ltfs_loans")

gold_master = (
    silver
    .withColumn("age_years", (months_between(col("disbursal_date"), col("date_of_birth")) / 12).cast("int"))
    .withColumn("default_status_label", when(col("is_default") == 1, "Default / NPA").otherwise("Standard Asset"))
    .withColumn("disbursed_amount_cr", sround(col("disbursed_amount") / 10000000.0, 4))
    .select(
        "loan_id", "branch_id", "supplier_id", "manufacturer_id",
        "state_id", "pincode_id", "Employee_code_ID",
        "disbursal_date", "employment_type", "age_years",
        "disbursed_amount", "disbursed_amount_cr", "asset_cost", "ltv", "ltv_band",
        "cibil_score", "cibil_description", "cibil_band",
        "aadhaar_flag", "PAN_flag", "VoterID_flag", "kyc_count",
        "pri_total_accts", "pri_active_accts", "pri_overdue_accts",
        "total_active_accts", "total_overdue_accts", "total_current_balance",
        "pri_instal_amt", "new_accts_6m", "delinquent_accts_6m",
        "avg_acct_age_months", "credit_history_months", "no_of_inquiries",
        "overdue_ratio", "is_default", "default_status_label"
    )
)

gold_master.write.mode("overwrite").format("delta").saveAsTable("gold_loan_master")
print(f"✅ Gold Master created successfully: {gold_master.count():,} rows")
```
