```python

from pyspark.sql.functions import current_timestamp, input_file_name, lit

# Read raw CSV file from OneLake Files
raw_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("Files/raw/train.csv")
)

```

```python
# Add metadata columns for data lineage
bronze_df = (
    raw_df
    .withColumn("_ingest_timestamp", current_timestamp())
    .withColumn("_source_file", input_file_name())
)
```

```python
# Save as Delta Table in Lakehouse
bronze_df.write.mode("overwrite").format("delta").saveAsTable("raw_ltfs_loans")

print(f"✅ Bronze layer loaded successfully. Total Records: {bronze_df.count():,}")
display(spark.sql("SELECT * FROM raw_ltfs_loans LIMIT 5"))
```
